# Step 1: Data Cleaning

## 1. Import Libraries

```python
from pathlib import Path

import numpy as np
import pandas as pd
```

## 2. Load Dataset

```python
paths = [
    Path("dataset/processed/student_performance_cleaned.csv.csv"),
    Path("../dataset/processed/student_performance_cleaned.csv.csv"),
    Path("dataset/processed/student_performance_cleaned.csv"),
    Path("../dataset/processed/student_performance_cleaned.csv"),
    Path(r"D:\adaptive-quiz-system\dataset\processed\student_performance_cleaned.csv.csv"),
]

dataset_path = next((p for p in paths if p.exists()), None)
if dataset_path is None:
    raise FileNotFoundError("Dataset not found")

df = pd.read_csv(dataset_path)
print("Dataset loaded:", dataset_path)
print("Shape:", df.shape)
```

## 4. Data Cleaning

```python
rows_before = len(df)

df = df.drop_duplicates()
df = df.dropna()
df["GradeClass"] = df["GradeClass"].astype(int)
df = df.reset_index(drop=True)

print("Rows before:", rows_before)
print("Rows after:", len(df))
```

## 5. Check After Cleaning

```python
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Shape:", df.shape)
```

## 6. Save Cleaned Dataset

```python
output_path = dataset_path.parent / "student_performance_final.csv"
df.to_csv(output_path, index=False)

print("Cleaned dataset saved:", output_path)
```
